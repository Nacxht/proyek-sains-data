# Separate Training Samples by Class

## Import Library

```python
import json
from pathlib import Path
```

## Load GeoJSON

```python
input_path = Path("digitizations/training_samples.geojson")
output_dir = Path("digitizations/separated")
output_dir.mkdir(parents=True, exist_ok=True)

with open(input_path) as f:
    data = json.load(f)

print(f"Total features: {len(data['features'])}")
```

## Group Features by Class

```python
classes: dict[str, list] = {}
for feature in data["features"]:
    props = feature["properties"]
    cls = props.get("Class") or props.get("class")
    classes.setdefault(cls, []).append(feature)

for cls, features in sorted(classes.items()):
    print(f"{cls}: {len(features)} features")
```

## Export Per-Class GeoJSON

```python
for cls, features in sorted(classes.items()):
    geojson = {
        "type": "FeatureCollection",
        "features": features,
    }
    out_path = output_dir / f"training_{cls}.geojson"
    with open(out_path, "w") as f:
        json.dump(geojson, f)
    print(f"Wrote {out_path} ({len(features)} features)")
```
