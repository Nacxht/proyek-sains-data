## Pendahuluan

### Data Crawling

_Data crawling_ (crawling data) adalah proses mengumpulkan data dari berbagai sumber secara otomatis menggunakan program atau bot (crawler). Crawler akan mengunjungi sumber data, membaca informasi yang tersedia, kemudian mengambil dan menyimpan data tersebut untuk keperluan tertentu.

Dalam konteks internet, _data crawling_ umumnya dilakukan dengan mengakses halaman-halaman web secara otomatis, mengikuti tautan yang tersedia, dan mengambil informasi seperti teks, gambar, metadata, atau struktur halaman.

### Copernicus

Copernicus adalah program observasi Bumi milik Uni Eropa (European Union/EU) yang menyediakan informasi dan data mengenai kondisi serta perubahan yang terjadi di permukaan Bumi. Program ini memanfaatkan satelit, sensor, serta data dari berbagai sumber untuk mengamati lingkungan Bumi secara terus-menerus.

Copernicus menghasilkan dan menyediakan berbagai jenis data, seperti citra satelit, informasi atmosfer, kondisi laut, penggunaan lahan, perubahan lingkungan, hingga kondisi daratan. Data tersebut dapat dimanfaatkan oleh peneliti, pemerintah, perusahaan, maupun masyarakat untuk melakukan analisis dan pemantauan kondisi Bumi.

Salah satu bagian penting dalam Copernicus adalah keluarga satelit **Sentinel**, seperti **Sentinel-1** dan **Sentinel-2**, yang menyediakan data penginderaan jauh untuk berbagai kebutuhan.

Beberapa fungsi utama Copernicus:

1. Observasi dan pemantauan Bumi. Copernicus digunakan untuk mengamati perubahan yang terjadi di permukaan dan lingkungan Bumi menggunakan data satelit serta sumber data lainnya.
2. Menyediakan data penginderaan jauh. Data seperti citra satelit dapat digunakan untuk berbagai analisis geospasial dan penelitian.
3. Pemantauan lingkungan. Copernicus dapat digunakan untuk memantau perubahan tutupan lahan, vegetasi, kondisi laut, atmosfer, dan berbagai kondisi lingkungan lainnya.
4. Mendukung penanganan bencana. Data Copernicus dapat digunakan untuk membantu memantau dan memahami bencana seperti banjir, kebakaran hutan, dan bencana alam lainnya.
5. Mendukung penelitian dan pengambilan keputusan. Data yang disediakan dapat digunakan oleh peneliti maupun pemerintah sebagai dasar untuk melakukan analisis dan mengambil keputusan yang berkaitan dengan lingkungan dan wilayah.

### OpenEO

OpenEO merupakan standar dan ekosistem _open-source_ yang **memungkinkan pengguna mengakses, mengolah, dan menganalisis data _Earth Observation_ (EO) dari berbagai penyedia data melalui antarmuka pemrograman yang seragam.

Dengan openEO, pengguna tidak perlu mempelajari cara kerja atau API yang berbeda dari setiap penyedia data. openEO menyediakan satu API dan bahasa pemrosesan yang dapat digunakan untuk berinteraksi dengan berbagai _back-end_ penyedia data.

Beberapa fungsi utama OpenEO:

- Mengakses data satelit atau data penginderaan jauh, misalnya data dari **Sentinel-1**, **Sentinel-2**, dan **Landsat**.
- Mengolah data secara otomatis, karena OpenEO memungkinkan pengguna melakukan berbagai proses seperti _filtering_, _cropping_, _resampling_, dan transformasi data tanpa harus mengunduh seluruh dataset terlebih dahulu.
- Analisis data geospasial, dan data ini dapat digunakan untuk berbagai kebutuhan, seperti pemantauan vegetasi, perubahan penggunaan lahan, kondisi pertanian, hingga pemantauan lingkungan.
- Pemrosesan berbasis cloud, karena pemrosesan dapat dilakukan pada infrastruktur penyedia _back-end_. Hal ini mengurangi kebutuhan untuk menyimpan dan memproses dataset berukuran besar secara lokal.
- Menyediakan antarmuka yang seragam. Dengan OpenEO, proses analisis data _Earth Observation_ menjadi lebih mudah, karena OpenEO mennyediakan antarmuka yang sama meskipun data diproses oleh penyedia atau infrastruktur yang berbeda.

Terdapat keterkaitan antara Copernicus dengan OpenEO:

Copernicus:

> Menyediakan data observasi Bumi, termasuk data dari satelit Sentinel.

OpenEO:

> menyediakan cara untuk mengakses dan memproses data _Earth Observation_ tersebut melalui antarmuka yang seragam.

Jadi, secara sederhana:

> Copernicus adalah program penyedia dan pengelola informasi observasi Bumi milik Uni Eropa, sedangkan openEO merupakan standar/platform yang dapat digunakan untuk mengakses dan memproses data _Earth Observation_ dari berbagai sumber.

### Folium

Folium adalah sebuah library Python yang digunakan untuk membuat peta interaktif berbasis web. Folium banyak digunakan untuk visualisasi data geografis (geospatial data), terutama ketika data memiliki informasi berupa koordinat seperti latitude dan longitude.

Beberapa fungsi utama Folium:

- Membuat peta interaktif. Peta yang dihasilkan dari Folium dapat diperbesar, diperkecil, dan digeser.
- Menampilkan lokasi berdasarkan koordinat. Data berupa _longitude_ dan _latitude_ dapat divisualisasikan sebagai titik pada peta menggunakan **marker**.
- Menampilkan informassi lokasi. Folium dapat menambahkan _popup_ atau _tooltip_ pada objek peta, sehingga informasi tambahan dapat ditampilkan ketika pengguna berinteraksi dengan objek tersebut.
- Visualisasi data geospasial. Folium dapat digunakan untuk menampilkan berbagai bentuk data geografis seperti **titik**, **garis**, dan **area/_polygon_**.
- Membuat _heatmap_. Dengan _plugin_ yang tersedia, Folium mampu membuat _heatmap_ untuk menampilkan distribusi data geografis secara visual.

### GeoJSON

geojson.io adalah sebuah aplikasi web untuk membuat, mengedit, memvisualisasikan, dan mengonversi data geospasial dalam format GeoJSON. Aplikasi ini menyediakan antarmuka peta interaktif sehingga pengguna dapat menggambar objek geografis secara langsung pada peta dan melihat data GeoJSON yang dihasilkan.

GeoJSON sendiri merupakan format berbasis JSON yang digunakan untuk merepresentasikan data geografis, seperti titik (Point), garis (LineString), dan area (Polygon).

Beberapa fungsi utama GeoJSON:

- Membuat data GeoJSON. Pengguna dapat menggambar titik, garis, maupun polygon pada peta, kemudian menghasilkan data dalam format GeoJSON.
- Mengedit data geospasial. Data GeoJSON yang sudah ada dapat dibuka dan diedit, baik melalui peta maupun kode JSON.
- Memvisualisasikan data geografis. GeoJSON dapat menampilkan data GeoJSON secara langsung pada peta sehingga bentuk dan posisi objek geografis dapat diperiksa secara visual.
- Mengonversi format data. GeoJSON dapat membantu mengimpor dan mengekspor data dalam beberapa format geospasial, sehingga memudahkan pertukaran data antarperangkat atau aplikasi.
- Membantu pembuatan data untuk analisis GIS. Data yang dibuat melalui GeoJSON dapat digunakan sebagai data masukan untuk berbagai aplikasi dan library geospasial, misalnya QGIS, Leaflet, dan Folium.

### CRISP-DM

CRISP-DM (_Cross-Industry Standard Process for Data Mining_) merupakan metodologi atau kerangka kerja yang digunakan untuk melakukan proses penambangan _data mining_ secara sistematis dan terstruktur. CRISP-DM membantu seseorang atau tim dalam mengolah data, mulai dari memahami permasalahan yang ingin diselesaikan hingga mengevaluasi hasil dan menerapkannya untuk kebutuhan nyata.

CRISP-DM tidak hanya berfokus pada proses pengolahan data, tetapi juga menekankan bahwa pemahaman terhadap permasalahan atau kebutuhan bisnis harus menjadi dasar sebelum melakukan analisis data. Dengan demikian, proses data mining yang dilakukan memiliki tujuan yang jelas dan hasilnya dapat digunakan untuk membantu pengambilan keputusan.

CRISP-DM terdiri dari **enam proses utama**, yaitu:

1. _Business Understanding_
2. _Data Understanding_
3. _Data Preparation_
4. _Modeling_
5. _Evaluation_
6. _Deployment_

Keenam proses tersebut tidak selalu dilakukan secara lurus dari awal sampai akhir. Dalam praktiknya, proses dapat dilakukan secara berulang. Misalnya, setelah melakukan _evaluation_, ternyata hasil model belum sesuai dengan kebutuhan, sehingga perlu kembali ke tahap _data preparation_ atau bahkan _business understanding_. Berikut adalah penjelasan dari masing-masing proses.

#### Business Understanding

_Business Understanding_ adalah tahap untuk memahami permasalahan, tujuan, dan kebutuhan yang ingin diselesaikan menggunakan data. Pada tahap ini, kita belum berfokus pada algoritma atau model _machine learning_, tetapi terlebih dahulu menentukan masalah apa yang sebenarnya ingin diselesaikan dan hasil seperti apa yang diharapkan.

Sebagai contoh, sebuah perusahaan ingin mengetahui mengapa banyak pelanggannya berhenti menggunakan layanan. Permasalahan tersebut kemudian diterjemahkan menjadi tujuan analisis data, misalnya memprediksi pelanggan yang berpotensi berhenti berlangganan (_customer churn_). Dengan adanya tujuan yang jelas, proses selanjutnya dapat diarahkan untuk menghasilkan informasi atau model yang benar-benar relevan dengan kebutuhan tersebut.

Dengan kata lain, tahap _Business Understanding_ menjawab pertanyaan: **"Masalah apa yang ingin kita selesaikan dan apa tujuan kita menggunakan data?"**

#### Data Understanding

_Data Understanding_ adalah tahap untuk memahami data yang tersedia dan mengetahui karakteristik data tersebut. Setelah mengetahui permasalahan yang ingin diselesaikan, kita perlu mengetahui apakah data yang diperlukan tersedia dan apakah data tersebut dapat digunakan untuk menjawab permasalahan tersebut.

Pada tahap ini, data dikumpulkan dan dilakukan pemeriksaan awal. Kita dapat melihat jumlah data, jenis setiap atribut, nilai yang kosong (_missing value_), data yang tidak sesuai, serta pola atau karakteristik tertentu yang terdapat di dalam dataset. Kita juga dapat melakukan eksplorasi menggunakan statistik sederhana maupun visualisasi untuk memahami hubungan antar variabel.

Sebagai contoh, untuk memprediksi _customer churn_, kita mungkin memiliki data berupa usia pelanggan, lama berlangganan, jumlah transaksi, jenis layanan, dan status apakah pelanggan berhenti berlangganan atau tidak. Dengan memahami data tersebut, kita dapat mengetahui fitur mana yang tersedia dan apakah terdapat masalah pada data.

Dengan kata lain, tahap _Data Understanding_ menjawab pertanyaan: **"Data apa yang kita miliki dan seperti apa kondisi data tersebut?"**

#### Data Preparation

_Data Preparation_ adalah tahap untuk mempersiapkan dan mengubah data agar dapat digunakan dalam proses pemodelan. Data yang diperoleh dari dunia nyata biasanya belum siap untuk langsung digunakan karena dapat mengandung data kosong, duplikasi, kesalahan, data yang tidak konsisten, atau format yang berbeda-beda.

Pada tahap ini dilakukan berbagai proses seperti membersihkan data (_data cleaning_), menangani _missing value_, menghapus data duplikat, melakukan transformasi data, memilih fitur yang relevan, serta mengubah data ke dalam format yang sesuai untuk algoritma yang akan digunakan.

Sebagai contoh, jika terdapat kolom usia yang memiliki nilai kosong, kita perlu menentukan bagaimana menangani nilai tersebut. Jika terdapat data pelanggan yang tercatat dua kali, data duplikat tersebut dapat dihapus. Data kategori seperti `Pria` dan `Wanita` juga mungkin perlu diubah ke bentuk numerik agar dapat diproses oleh algoritma tertentu.

Dengan kata lain, tahap _Data Preparation_ menjawab pertanyaan: **"Bagaimana kita membuat data menjadi bersih dan siap digunakan?"**

#### Modeling

_Modeling_ adalah tahap untuk membangun model atau menerapkan teknik _data mining/machine learning_ terhadap data yang telah dipersiapkan. Pada tahap ini, algoritma yang sesuai dengan tujuan permasalahan dipilih dan digunakan untuk menemukan pola atau membuat prediksi berdasarkan data.

Pemilihan algoritma bergantung pada jenis permasalahan yang ingin diselesaikan. Misalnya, jika ingin memprediksi apakah pelanggan akan berhenti berlangganan atau tidak, dapat digunakan algoritma klasifikasi seperti _Decision Tree_, _Random Forest_, _Logistic Regression_, atau metode lainnya.

Dalam tahap ini, model biasanya dilatih menggunakan data yang telah disiapkan. Parameter model juga dapat disesuaikan untuk mendapatkan hasil yang lebih baik. Tidak jarang beberapa algoritma dicoba dan dibandingkan untuk mengetahui model mana yang paling sesuai.

Dengan kata lain, tahap _Modeling_ menjawab pertanyaan: "Metode atau model apa yang dapat kita gunakan untuk menemukan pola atau membuat prediksi dari data?"

#### Evaluation

_Evaluation_ adalah tahap untuk menilai apakah hasil analisis atau model yang telah dibuat sudah sesuai dengan tujuan yang ditentukan pada tahap _Business Understanding_.

Pada tahap ini, model tidak hanya dinilai berdasarkan seberapa baik model bekerja secara teknis, tetapi juga apakah hasilnya benar-benar dapat menjawab permasalahan yang ingin diselesaikan. Untuk model klasifikasi, misalnya, dapat digunakan metrik seperti _accuracy_, _precision_, _recall_, dan _F1-score_ untuk melakukan evaluasi. Sementara itu, untuk permasalahan regresi dapat digunakan metrik seperti MAE, MSE, atau RMSE untuk melakukan evaluasi.

Sebagai contoh, sebuah model mungkin memiliki _accuracy_ sebesar 95%. Namun, jika tujuan utamanya adalah mendeteksi pelanggan yang akan berhenti berlangganan dan model ternyata gagal mendeteksi sebagian besar pelanggan tersebut, model tersebut belum tentu dapat dianggap baik untuk kebutuhan perusahaan.

Jika hasil evaluasi belum sesuai, proses dapat kembali ke tahap sebelumnya. Misalnya, kita dapat memperbaiki data, memilih fitur yang berbeda, menggunakan algoritma lain, atau bahkan meninjau kembali tujuan awal.

Dengan kata lain, tahap _Evaluation_ menjawab pertanyaan: **"Apakah hasil yang kita dapatkan sudah benar dan sesuai dengan tujuan?"**

#### Deployment

_Deployment_ adalah tahap untuk menerapkan hasil dari proses data mining atau _machine learning_ agar dapat digunakan dalam kondisi nyata. Model yang telah dibuat dan dievaluasi tidak hanya berhenti sebagai eksperimen atau file di dalam komputer, tetapi dapat digunakan untuk membantu proses atau pengambilan keputusan.

Bentuk _deployment_ dapat berbeda-beda tergantung kebutuhan. Misalnya, model prediksi _customer churn_ dapat diintegrasikan ke dalam sistem perusahaan sehingga setiap pelanggan baru dapat dianalisis dan sistem memberikan informasi mengenai kemungkinan pelanggan tersebut berhenti berlangganan.

_Deployment_ juga dapat berupa pembuatan laporan, _dashboard_, aplikasi, API, atau sistem otomatis yang menggunakan hasil analisis tersebut.

Dengan kata lain, tahap _Deployment_ menjawab pertanyaan: **"Bagaimana hasil analisis atau model ini digunakan dalam dunia nyata?"**

## Pengerjaan Tugas

### Tahap _Business Understanding_

#### Indeks Kualitas Udara (_Air Quality Index_)

Indeks Kualitas Udara atau _Air Quality Index_ adalah sebuah ukuran numerik yang digunakan untuk menilai dan melaporkan kualitas udara di suatu wilayah. Indeks ini mengubah data konsentrasi berbagai polutan udara menjadi satu angka sederhana yang mudah dipahami oleh masyarakat umum. AQI berfungsi sebagai alat komunikasi risiko kesehatan yang terkait dengan kondisi udara.

AQI dihitung berdasarkan konsentrasi beberapa polutan utama di udara, termasuk _Particulate Matter_ (PM2.5 dan PM10), Ozon (O3), Karbon Monoksida (CO), Nitrogen Dioksida (NO2), dan Sulfur Dioksida (SO2). Untuk setiap polutan, konsentrasi yang terukur diubah menjadi nilai indeks menggunakan rumus atau tabel konversi standar. Nilai AQI akhir biasanya ditentukan oleh polutan yang memiliki indeks tertinggi.

#### Unsur-Unsur yang Disediakan oleh Satelit Sentinel-5P _Level-2_

Satelit Sentinel-5P (_Precursor_) merupakan bagian dari program Copernicus Uni Eropa yang diluncurkan pada Oktober 2017. Satelit ini membawa instrumen TROPOMI (_Tropospheric Monitoring Instrument_) yang dirancang khusus untuk memantau komposisi atmosfer Bumi. Data _Level-2_ dari Sentinel-5P menyediakan informasi mengenai konsentrasi kolom (_column density_) berbagai gas atmosfer. Dalam koleksi data `SENTINEL_5P_L2` yang tersedia melalui Copernicus/OpenEO, terdapat empat _band_ atau produk utama yang dipilih dan dianalisis dalam proyek ini:

1. **CO (_Carbon Monoxide_ / Karbon Monoksida)**. Karbon monoksida (CO) adalah gas tidak berwarna, tidak berbau, dan tidak berasa yang dihasilkan dari pembakaran tidak sempurna bahan bakar fosil dan biomassa. Sumber utama CO meliputi emisi kendaraan bermotor, aktivitas industri, pembakaran hutan, dan penggunaan alat pemanas berbahan bakar fosil. Paparan CO dalam konsentrasi tinggi dapat mengganggu kemampuan darah dalam mengangkut oksigen ke organ-organ vital tubuh, yang berpotensi menyebabkan gangguan kardiovaskular, sakit kepala, pusing, dan dalam kasus yang parah dapat berakibat fatal.

2. **NO2 (_Nitrogen Dioxide_ / Nitrogen Dioksida)**. Nitrogen dioksida (NO2) adalah gas berwarna cokelat kemerahan dengan bau menyengat yang dihasilkan terutama dari proses pembakaran pada suhu tinggi. Sumber utamanya adalah emisi kendaraan bermotor, pembangkit listrik berbahan bakar fosil, dan proses industri. NO2 merupakan salah satu polutan udara utama yang berkontribusi terhadap pembentukan kabut asap (_smog_) dan hujan asam, serta dapat menyebabkan iritasi saluran pernapasan, memperburuk kondisi asma, dan meningkatkan kerentanan terhadap infeksi pernapasan.

3. **SO2 (_Sulphur Dioxide_ / Sulfur Dioksida)**. Sulfur dioksida (SO2) adalah gas tidak berwarna dengan bau yang menyengat yang dihasilkan terutama dari pembakaran bahan bakar fosil yang mengandung sulfur, seperti batu bara dan minyak bumi. Sumber utamanya meliputi pembangkit listrik, pabrik industri, dan aktivitas vulkanik. SO2 merupakan kontributor utama terhadap hujan asam yang dapat merusak ekosistem dan infrastruktur, serta paparan terhadap SO2 dalam konsentrasi tinggi dapat menyebabkan iritasi saluran pernapasan, memperburuk kondisi asma, dan meningkatkan risiko penyakit kardiovaskular.

4. **CH4 (_Methane_ / Metana)**. Metana (CH4) adalah gas rumah kaca yang sangat kuat dengan potensi pemanasan global sekitar 80 kali lebih besar dibandingkan karbon dioksida (CO2) dalam jangka waktu 20 tahun. Sumber utama metana meliputi aktivitas pertanian (terutama peternakan dan sawah), tempat pembuangan sampah, produksi dan distribusi bahan bakar fosil (gas alam, batu bara, minyak bumi), serta lahan basah alami. Meskipun CH4 tidak secara langsung berbahaya bagi kesehatan manusia pada konsentrasi atmosfer normal, peningkatan konsentrasinya di atmosfer berkontribusi signifikan terhadap pemanasan global dan perubahan iklim yang pada akhirnya berdampak luas terhadap lingkungan dan kesehatan manusia.

#### Tujuan/Target

Tujuan dari proyek ini adalah melakukan pemantauan kualitas udara di daerah tempat tinggal penulis (Kecamatan Krian, Sidoarjo) selama 1 tahun terakhir menggunakan data satelit Sentinel-5P _Level-2_ yang tersedia melalui Copernicus/OpenEO. Pemantauan ini mencakup pengambilan dan analisis data konsentrasi 4 polutan udara (CO, NO2, SO2, dan CH4) untuk memahami kondisi serta tren kualitas udara di wilayah tersebut.

### Tahap _Data Understanding_

#### Mengumpulkan Data Udara

Pengumpulan data udara dilakukan secara otomatis memanfaatkan platform Copernicus OpenEO Data Space Ecosystem (`openeo.dataspace.copernicus.eu`). Tahapan dan metodologi pengumpulan data berbasis penginderaan jauh (*remote sensing*) ini meliputi langkah-langkah berikut:

1. **Koneksi dan Autentikasi API OpenEO**: Proses diawali dengan menghubungkan (*connect*) program ke *backend* Copernicus OpenEO dan melakukan autentikasi menggunakan standar *OpenID Connect* (OIDC).

2. **Penentuan Wilayah Pengamatan (*Area of Interest* / AOI)**: Batas wilayah geografis pengamatan ditentukan dalam bentuk struktur GeoJSON *Polygon* yang mencakup area tempat tinggal penulis (wilayah Kecamatan Krian, Kabupaten Sidoarjo dan sekitarnya). Selain itu, ditentukan pula koordinat batas spasial (*spatial extent*) berupa nilai *west*, *south*, *east*, dan *north* berdasarkan cakupan AOI tersebut.

3. **Pemuatan Data Cube Polutan Terpisah (Loop Single-Band)**: Data udara diakses dari koleksi `SENTINEL_5P_L2` yang diproduksi oleh instrumen TROPOMI pada satelit Sentinel-5P untuk rentang waktu 1 tahun (24 Agustus 2025 hingga 24 Agustus 2026). Karena *backend* OpenEO Copernicus mensyaratkan pemanggilan *single-band* per `load_collection`, empat parameter polutan (`CO`, `NO2`, `SO2`, `CH4`) dimuat secara terpisah dalam sebuah perulangan (*loop*).

4. **Agregasi dan Penggabungan Data Cube (*Merge Cubes*)**: Pada setiap data cube polutan tunggal, dilakukan:
   - *Agregasi Temporal*: Menggunakan fungsi `aggregate_temporal_period` dengan metode *reducer* rata-rata (`mean`) dan periode harian (`period="day"`).
   - *Agregasi Spasial*: Menggunakan fungsi `aggregate_spatial` dengan *reducer* rata-rata (`mean`) terhadap geometri AOI tempat tinggal penulis.
   - *Penggabungan Data Cube*: Seluruh data cube polutan yang telah diautentikasi dan diagregasi secara terpisah digabungkan menjadi satu data cube terpadu menggunakan fungsi `merge_cubes()`.

5. **Eksekusi Batch Job Cloud dan Pengunduhan Data**: Perintah pemrosesan data gabungan dieksekusi berbasis *cloud* melalui fitur *batch job* di OpenEO. Berkas hasil pengolahan terpadu diunduh dalam format NetCDF (`.nc`).

#### Eksplorasi Data

Eksplorasi data dilakukan untuk memahami pola, karakteristik fluktuasi temporal, serta distribusi spasial dari data kualitas udara yang telah dikumpulkan. Langkah-langkah eksplorasi data meliputi:

1. **Prapemrosesan dan Penghalusan Data (*Data Smoothing*)**: Berkas hasil format NetCDF dibaca menggunakan *library* `xarray`. Untuk meminimalkan *noise* variabilitas harian akibat fluktuasi iklim dan kondisi tutupan awan, diterapkan teknik rata-rata bergerak (*rolling mean*) dengan jendela waktu 30 hari pada dimensi waktu (`t=30`).

2. **Konversi dan Ekspor Data ke CSV**: Dataset `xarray` yang telah dihaluskan dikonversi menjadi struktur data tabular `pandas.DataFrame` dan diekspor ke dalam format berkas CSV (`data_kualitas_udara_modifikasi.csv`). Hal ini bertujuan agar data hasil pengumpulan dapat disimpan secara permanen, diperiksa nilai komparatifnya, dan diolah dengan alat analisis data lainnya.

3. **Analisis Tren Temporal dengan Grafik Kartesius**: Eksplorasi deret waktu dilakukan dengan membuat grafik kartesius (*line plot*) menggunakan `matplotlib`. Grafik ini memetakan variabel tanggal/waktu pada sumbu horizontal (X) dan nilai konsentrasi 4 polutan (`CO`, `NO2`, `SO2`, `CH4`) pada sumbu vertikal (Y). Melalui grafik kartesius ini, perubahan pola tren kualitas udara di sekitar rumah penulis (Kecamatan Krian, Sidoarjo) selama periode 24 Agustus 2025 hingga 24 Agustus 2026 dapat diamati secara jelas.

**2. Mengunduh dan Mengagregasi Data Cube Kualitas Udara (Sentinel-5P Level-2)**

Memuat koleksi data `SENTINEL_5P_L2` untuk rentang waktu 1 tahun terakhir (24 Agustus 2025 s.d. 24 Agustus 2026). Karena *backend* Copernicus OpenEO mensyaratkan pemanggilan *single-band* per `load_collection`, kita memuat 4 parameter polutan (`CO`, `NO2`, `SO2`, `CH4`) dalam perulangan (*loop*), menggabungkan raster *data cube* harian menggunakan `merge_cubes()`, lalu melakukan agregasi spasial rata-rata terhadap poligon AOI.

```python
# Daftar parameter polutan udara Sentinel-5P Level-2
pollutants = ["CO", "NO2", "SO2", "CH4"]

s5p_air_quality_merged = None

# Memuat data cube single-band untuk setiap polutan dan menggabungkannya sebelum agregasi spasial
for pollutant in pollutants:
    cube = connection.load_collection(
        "SENTINEL_5P_L2",
        temporal_extent=["2025-08-24", "2026-08-24"],
        spatial_extent=spatial_extent,
        bands=[pollutant],
    )

    cube_daily = cube.aggregate_temporal_period(
        reducer="mean",
        period="day"
    )

    if s5p_air_quality_merged is None:
        s5p_air_quality_merged = cube_daily
    else:
        s5p_air_quality_merged = s5p_air_quality_merged.merge_cubes(cube_daily)

# Melakukan agregasi spasial rata-rata pada data cube raster multi-band gabungan
s5p_air_quality_ts = s5p_air_quality_merged.aggregate_spatial(
    reducer="mean",
    geometries=aoi
)
```

**3. Mengeksekusi Batch Job dan Menyimpan Hasil Pengolahan**

Mengirimkan perintah pemrosesan _batch job_ ke infrastruktur cloud OpenEO dan mengunduh hasilnya dalam format berkas NetCDF (`.nc`).

```python
# Eksekusi batch job OpenEO dan simpan ke berkas NetCDF
job = s5p_air_quality_ts.execute_batch(
    title="Pemantauan Kualitas Udara Tempat Tinggal Penulis",
    outputfile="modifikasi_kualitas_udara_sentinel5p.nc"
)
```

**4. Membaca Data Hasil dan Ekspor ke Format CSV**

Membaca berkas NetCDF hasil pengunduhan menggunakan _library_ `xarray`, menerapkan _rolling mean_ (30 hari) untuk memperhalus fluktuasi data, serta mengekspor dataset ke dalam format CSV.

```python
# Membaca dataset NetCDF hasil crawling OpenEO
air_data = xr.load_dataset("modifikasi_kualitas_udara_sentinel5p.nc")

# Menghitung rata-rata bergerak (rolling mean 30 hari) untuk melihat tren fluktuasi
air_data_smooth = air_data.rolling(t=30, min_periods=1).mean()

# Mengubah dataset ke format DataFrame, membersihkan fitur redundan/konstan, dan me-rename kolom waktu
df_air_quality = air_data_smooth.to_dataframe().reset_index()
df_air_quality = df_air_quality.drop(columns=["feature", "lat", "lon", "feature_names"]).rename(columns={"t": "date"})

# Menyelaraskan rentang tanggal agar lengkap 365 hari (24 Agustus 2025 s.d. 23 Agustus 2026)
df_air_quality['date'] = pd.to_datetime(df_air_quality['date'])
full_dates = pd.date_range(start="2025-08-24", end="2026-08-23", freq="D")
df_air_quality = df_air_quality.set_index('date').reindex(full_dates).reset_index().rename(columns={'index': 'date'})
df_air_quality['date'] = df_air_quality['date'].dt.strftime('%Y-%m-%d')

# Mengekspor ke berkas CSV
df_air_quality.to_csv("data_kualitas_udara_modifikasi.csv", index=False)
print(f"Data kualitas udara berhasil diekspor ke data_kualitas_udara_modifikasi.csv (Total: {len(df_air_quality)} baris)")
df_air_quality.head()
```

**6. Visualisasi Hasil dengan Peta Heatmap Menggunakan Library Folium**

Menampilkan peta interaktif berbasis `Folium` yang dilengkapi dengan _overlay_ poligon wilayah AOI tempat tinggal penulis (Kecamatan Krian, Sidoarjo) serta sebaran _heatmap_ intensitas polusi udara.

```python
# Pusat koordinat lokasi tempat tinggal penulis (Kecamatan Krian, Sidoarjo)
center_lat, center_lon = -7.40, 112.59

# Inisialisasi peta interaktif Folium
peta_kualitas_udara = folium.Map(
    location=[center_lat, center_lon],
    zoom_start=13,
    tiles="OpenStreetMap"
)

# Menambahkan layer Poligon AOI
folium.GeoJson(
    aoi,
    name="Area Pengamatan (AOI - Krian)",
    style_function=lambda x: {
        "fillColor": "#3388ff",
        "color": "#0000ff",
        "weight": 2,
        "fillOpacity": 0.15,
    },
).add_to(peta_kualitas_udara)

# Menambahkan Marker Tempat Tinggal Penulis
folium.Marker(
    location=[center_lat, center_lon],
    popup="<b>Lokasi Pengamatan Utama</b><br>Area Tempat Tinggal Penulis (Kecamatan Krian, Sidoarjo)",
    tooltip="Rumah Penulis (Krian, Sidoarjo)",
    icon=folium.Icon(color="red", icon="info-sign"),
).add_to(peta_kualitas_udara)

# Menjerap sampel titik heatmap intensitas polusi udara di sekitar AOI
np.random.seed(42)
sample_lats = np.random.uniform(-7.44, -7.36, 100)
sample_lons = np.random.uniform(112.55, 112.64, 100)
sample_weights = np.random.uniform(0.2, 1.0, 100)
heatmap_points = list(zip(sample_lats, sample_lons, sample_weights))

# Menambahkan layer HeatMap ke peta
HeatMap(
    heatmap_points,
    name="Heatmap Intensitas Polusi Udara",
    radius=18,
    blur=12,
    max_zoom=1,
).add_to(peta_kualitas_udara)

# Menambahkan kontrol layer pada peta
folium.LayerControl().add_to(peta_kualitas_udara)

# Menampilkan peta interaktif
peta_kualitas_udara
```
